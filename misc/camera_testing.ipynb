```python
import pyrealsense2 as rs
import numpy as np
import cv2

pipeline = rs.pipeline()
config = rs.config()

# Enable color stream
config.enable_stream(rs.stream.color, 640, 480, rs.format.bgr8, 30)

pipeline.start(config)
print("RealSense started. Press Q to quit.")

try:
    while True:
        frames = pipeline.wait_for_frames()
        color_frame = frames.get_color_frame()

        if not color_frame:
            print("⚠️ No frame received")
            continue

        frame = np.asanyarray(color_frame.get_data())

        cv2.imshow("RealSense", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
finally:
    pipeline.stop()
    cv2.destroyAllWindows()
    print("Done.")
```

```python
import pyrealsense2 as rs
import numpy as np
import cv2

# First, detect connected RealSense devices
ctx = rs.context()
devices = ctx.query_devices()

if len(devices) == 0:
    print("❌ No RealSense devices found")
else:
    for i, dev in enumerate(devices):
        print(f"✅ Device {i}: {dev.get_info(rs.camera_info.name)}")
        print(f"   Serial: {dev.get_info(rs.camera_info.serial_number)}")
        print(f"   USB: {dev.get_info(rs.camera_info.usb_type_descriptor)}")

pipeline = rs.pipeline()
config = rs.config()

# Try lower resolution first — USB 2.0 ports struggle with 640x480 @ 30fps
config.enable_stream(rs.stream.color, 424, 240, rs.format.bgr8, 6)

try:
    profile = pipeline.start(config)
    dev = profile.get_device()
    print(f"\nUsing: {dev.get_info(rs.camera_info.name)}")
    print("Streaming... Press Q to quit.")

    # Warm up — discard first few frames
    for _ in range(30):
        pipeline.wait_for_frames(timeout_ms=10000)

    while True:
        frames = pipeline.wait_for_frames(timeout_ms=10000)
        color_frame = frames.get_color_frame()

        if not color_frame:
            continue

        frame = np.asanyarray(color_frame.get_data())
        cv2.imshow("RealSense", frame)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

except RuntimeError as e:
    print(f"❌ Error: {e}")
    print("\nTry:")
    print("  1. Plug into a USB 3.0 port (blue port)")
    print("  2. Use a different USB cable")
    print("  3. Restart the RealSense Viewer app first, then close it and re-run")

finally:
    pipeline.stop()
    cv2.destroyAllWindows()
```

```python
import cv2

# Try each one and see which gives a clean image
backends = [
    (cv2.CAP_DSHOW, "DirectShow"),
    (cv2.CAP_MSMF, "Media Foundation"),
    (cv2.CAP_ANY, "Auto"),
]

for backend, name in backends:
    cap = cv2.VideoCapture(2, backend)
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)
    
    ret, frame = cap.read()
    print(f"{name}: ret={ret}, shape={frame.shape if ret else 'N/A'}")
    
    if ret and frame is not None:
        cv2.imshow(f"Test - {name}", frame)
        cv2.waitKey(2000)
    
    cap.release()
    cv2.destroyAllWindows()
```
